## Functions to create probability distribution 
### Binomial distribution

Mathematical expression for bionomial distribution
$$
 P(X=k) = \binom{n}{k}p^k(1-p)^{n-k}
$$

Insert a figure

P = probability of having k successes out of n

$\binom{n}{k}$ = binomial coeffcient

n = number of trials


k = number of successes (0 ≤ x ≤ n)


p = probability of success in a single trial


(1−p) = probability of failure in a single trial

more about [math experssions](https://jupyterbook.org/en/stable/content/math.html)


- **Yes-no** questions for n number of **independent** experiments with constant sucess **probability**
