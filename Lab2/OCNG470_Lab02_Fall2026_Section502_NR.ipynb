## Part 2A: Binomial Distribution

```python
n = 25
p = .5
k = 7

manual_prob = factorial(n) / (factorial(k) * factorial(n-k)) * p**k * (1-p)**(n-k)
python_prob = binom.pmf(k, n, p)

print(f"Manual Probability: {manual_prob}")
print(f"Python Probability: {python_prob}")

X = np.arange(0, 26)
pdf_fair = binom.pmf(X, n, p)
# X, pdf_fair

plt.figure(figsize=(10, 8))
plt.subplot(2, 1, 1)
plt.bar(X, pdf_fair)
plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.title("Binomial Distribution: n = 25, p = 0.5")
plt.xticks(X)
plt.tight_layout()
# plt.show()
plt.close()

p_biased = 0.8
pdf_biased = binom.pmf(X, n, p_biased)

plt.figure(figsize=(10, 8))
plt.subplot(2, 1, 1)
plt.bar(X, pdf_fair)
plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.title("Binomial Distribution: n = 25, p = 0.5")
plt.xticks(X)
plt.subplot(2, 1, 2)
plt.bar(X, pdf_biased)
plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.title("Binomial Distribution: n = 25, p = 0.8")
plt.xticks(X)
plt.tight_layout()
plt.show()
```

## Part 2B: Poisson and Normal Distributions

```python
Lambda = [1, 4, 10, 15]
x = 4
for lam in Lambda:
    # Calculate Poisson probability manually
    calc_prob = lam**x * exp(-lam) / factorial(x)
  
    # Check with scipy.stats.poisson.pmf()
    scipy_prob = poisson.pmf(x, lam)
  
    print(f"P(x = 4, lambda = {lam}) manually = {calc_prob}")
    print(f"P(x = 4, lambda = {lam}) scipy    = {scipy_prob}")
    print()

X = np.arange(0, 31)
Lambda = [1, 4, 10, 15]
plt.figure(figsize=(12, 8))
for i, lam in enumerate(Lambda):
    poisson_pdf = poisson.pmf(X, lam)
    plt.subplot(2, 2, i + 1)
    plt.bar(X, poisson_pdf)
    plt.xlabel("X")
    plt.ylabel("Probability")
    plt.title(f"Poisson Distribution: lambda = {lam}")
plt.tight_layout()
plt.show()

X = np.arange(0, 101)
mu = 25
sigma_values = [5, 10]
plt.figure(figsize=(10, 8))
for i, sigma in enumerate(sigma_values):
        # Calculate normal PDF manually
    normal_pdf = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((X - mu) / sigma)**2)
    plt.subplot(2, 1, i + 1)
    plt.bar(X, normal_pdf)
    plt.xlabel("X")
    plt.ylabel("Probability Density")
    plt.title(f"Normal Distribution: mu = {mu}, sigma = {sigma}")
plt.tight_layout()
plt.show()
```

The probability of rolling exactly two "1"s in five rolls is 0.263671875 or ~26%

The probability of rolling exactly two consecutive "1"s in five rolls is 0.10546875 OR ~11%

The probability of rolling 2 ones is higher than the probability of two consecutive ones as rolling two ones encompasses all of the possible outcomes of rolling two consecutive ones and also contains the possible outcomes of two non-consecutive ones
