```python
# Lab 9 Part A: One-Way ANOVA Analysis
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd
import warnings
warnings.filterwarnings('ignore')

sns.set_style("whitegrid")
np.random.seed(42)

# Generate exponential data
x = np.exp(np.arange(1, 31) / 10)
noise_factor = 1
data_ns1 = np.zeros((10, 30))

for i in range(10):
    noise = np.random.randn(30) * noise_factor
    data_ns1[i, :] = x + noise
```

```python
# Figure 1: 10 vs 30 treatment groups
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

for i in range(10):
    ax1.plot(data_ns1[i, :], alpha=0.7, linewidth=1.5)
ax1.set_title('10 treatment groups', fontweight='bold')
ax1.set_xlabel('Sample ID')
ax1.set_ylabel('Measured values')

for i in range(30):
    ax2.plot(data_ns1[:, i], alpha=0.7, linewidth=1.5)
ax2.set_title('30 treatment groups', fontweight='bold')
ax2.set_xlabel('Sample ID')
ax2.set_ylabel('Measured values')

plt.tight_layout()
plt.show()
```

```python
# One-way ANOVA: 10 treatment groups
data_10 = []
for group in range(10):
    for value in data_ns1[group, :]:
        data_10.append({'Treatment': f'Group_{group+1}', 'Value': value})
df_10 = pd.DataFrame(data_10)

groups_10 = [df_10[df_10['Treatment'] == f'Group_{i+1}']['Value'].values for i in range(10)]
f_stat_10, p_val_10 = stats.f_oneway(*groups_10)
print(f"10 Groups: F={f_stat_10:.4f}, p={p_val_10:.6e}")

# Box plot
fig, ax = plt.subplots(figsize=(12, 6))
df_10.boxplot(column='Value', by='Treatment', ax=ax)
plt.suptitle('')
ax.set_title('ANOVA1: 10 treatments, Noise_factor = 1', fontweight='bold')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Multiple comparison
tukey_10 = pairwise_tukeyhsd(df_10['Value'], df_10['Treatment'], alpha=0.05)
fig, ax = plt.subplots(figsize=(10, 8))
tukey_10.plot_simultaneous(ax=ax)
ax.set_title('Multcompare: 10 treatments, Noise_factor = 1', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# One-way ANOVA: 30 treatment groups
data_30 = []
for group in range(30):
    for value in data_ns1[:, group]:
        data_30.append({'Treatment': f'Sample_{group+1}', 'Value': value})
df_30 = pd.DataFrame(data_30)

groups_30 = [df_30[df_30['Treatment'] == f'Sample_{i+1}']['Value'].values for i in range(30)]
f_stat_30, p_val_30 = stats.f_oneway(*groups_30)
print(f"30 Groups: F={f_stat_30:.4f}, p={p_val_30:.6e}")

# Box plot
fig, ax = plt.subplots(figsize=(16, 6))
df_30.boxplot(column='Value', by='Treatment', ax=ax)
plt.suptitle('')
ax.set_title('ANOVA1: 30 treatments, Noise_factor = 1', fontweight='bold')
plt.xticks(rotation=90, fontsize=8)
plt.tight_layout()
plt.show()

# Multiple comparison
tukey_30 = pairwise_tukeyhsd(df_30['Value'], df_30['Treatment'], alpha=0.05)
fig, ax = plt.subplots(figsize=(12, 10))
tukey_30.plot_simultaneous(ax=ax)
ax.set_title('Multcompare: 30 treatments, Noise_factor = 1', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Lab 9 Part B: Two-Way ANOVA Analysis
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from statsmodels.formula.api import ols

sns.set_style("whitegrid")
sns.set_palette("husl")

# Set random seed for reproducibility (optional - remove if you want different results each time)
np.random.seed(42)

a_levels = np.arange(1, 5)
b_levels = np.arange(1, 5)

def run_anova_case(amesh, bmesh, case_name, case_description):
    """Run ANOVA analysis and create plots for a given case"""
    # Generate data with 10 replicates
    n_reps = 10
    data_list = []
    
    for rep in range(n_reps):
        for i in range(4):
            for j in range(4):
                response = amesh[i, j] + bmesh[i, j] + np.random.rand() - 0.5
                data_list.append({
                    'Factor1': i + 1,
                    'Factor2': j + 1,
                    'Response': response,
                    'Replicate': rep
                })
    
    data = pd.DataFrame(data_list)
    
    # Create plots
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Raw data plot - plot by Factor2 groups
    for j in range(1, 5):
        subset = data[data['Factor2'] == j]
        ax1.plot(range(len(subset)), subset['Response'].values, 
                alpha=0.7, linewidth=1.5, label=f'Factor2={j}')
    ax1.set_title(f'{case_name}: Raw Data\n{case_description}', fontweight='bold', fontsize=11)
    ax1.set_xlabel('Observation Number')
    ax1.set_ylabel('Response Value')
    ax1.legend(title='Factor 2', fontsize=9)
    
    # Interaction plot using seaborn
    sns.pointplot(data=data, x='Factor1', y='Response', hue='Factor2', 
                  ax=ax2, markers='o', linestyles='-', errorbar='se', err_kws={'linewidth': 1})
    ax2.set_title(f'{case_name}: Mean Response by Factor Levels', fontweight='bold', fontsize=11)
    ax2.set_xlabel('Factor 1 Levels')
    ax2.set_ylabel('Mean Response')
    ax2.legend(title='Factor 2', loc='best', fontsize=9)
    
    plt.tight_layout()
    plt.show()
    
    # ANOVA table
    model = ols('Response ~ C(Factor1) + C(Factor2) + C(Factor1):C(Factor2)', data=data).fit()
    anova_table = sm.stats.anova_lm(model, typ=2)
    
    print(f'\n{"="*75}')
    print(f'ANOVA Table - {case_name}: {case_description}')
    print("="*75)
    print(anova_table.to_string())
    
    # Interpretation helper
    print(f'\n{"Interpretation:":-^75}')
    for idx in anova_table.index[:-1]:  # Skip 'Residual'
        pval = anova_table.loc[idx, 'PR(>F)']
        sig = "✓ SIGNIFICANT" if pval < 0.05 else "✗ NOT significant"
        print(f"  {idx:30s}: p-value = {pval:.6f}  {sig}")
    print("="*75 + "\n")
    
    return anova_table, data
```

```python
## Case 1: Both factors have linear effects
print("\n" + "🔵 CASE 1: TWO LINEAR FACTORS".center(75, "="))
amesh1, bmesh1 = np.meshgrid(a_levels, b_levels, indexing='ij')
anova1, data1 = run_anova_case(amesh1, bmesh1, 'Case 1', 
                                'Two Linear Factors (Response = F1 + F2 + noise)')
```

```python
## Case 2: One linear factor, one random factor
print("\n" + "🟡 CASE 2: ONE LINEAR + ONE RANDOM FACTOR".center(75, "="))
amesh2, _ = np.meshgrid(a_levels, b_levels, indexing='ij')
bmesh2 = np.random.rand(4, 4)
anova2, data2 = run_anova_case(amesh2, bmesh2, 'Case 2', 
                                'Linear + Random (Response = F1 + random + noise)')
```

```python
## Case 3: Both factors are random
print("\n" + "🟠 CASE 3: TWO RANDOM FACTORS".center(75, "="))
amesh3 = np.random.rand(4, 4)
bmesh3 = np.random.rand(4, 4)
anova3, data3 = run_anova_case(amesh3, bmesh3, 'Case 3', 
                                'Two Random Factors (Response = random1 + random2 + noise)')
```

```python
## Case 4: One exponential factor, one linear factor
print("\n" + "🔴 CASE 4: EXPONENTIAL + LINEAR FACTOR".center(75, "="))
amesh4 = np.meshgrid(a_levels, b_levels, indexing='ij')[0] ** 2
bmesh4 = np.meshgrid(a_levels, b_levels, indexing='ij')[1]
anova4, data4 = run_anova_case(amesh4, bmesh4, 'Case 4', 
                                'Exponential + Linear (Response = F1² + F2 + noise)')
```
