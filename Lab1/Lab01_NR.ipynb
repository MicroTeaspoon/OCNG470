```python
# ---- Part 1C: Importing and visualizing NetCDF variables ----
from netCDF4 import Dataset
import numpy as np
import matplotlib.pyplot as plt
nc_file = "R4901473_154.nc"
# Open NetCDF file
nc = Dataset(nc_file, mode="r")
# Read temperature and pressure variables
temp = nc.variables["TEMP"][:]
pres = nc.variables["PRES"][:]
# Close NetCDF file
nc.close()
# Convert to one-dimensional arrays
temp = np.array(temp[0]).flatten()
pres = np.array(pres[0]).flatten()
# Remove missing or invalid values
good = np.isfinite(temp) & np.isfinite(pres)
temp = temp[good]
pres = pres[good]
# Plot temperature versus pressure
plt.figure()
plt.plot(temp, pres, color="blue", linewidth=2)
plt.xlabel("Temperature (°C)")
plt.ylabel("Pressure (dbar)")
plt.title("Temperature vs Pressure")
plt.gca().invert_yaxis()
plt.grid(True)
plt.savefig("OCNG470_Lab01_Profile.pdf")
plt.close()
```

```python
# ---- Part 1D: Importing and visualizing text data ----
import numpy as np
import matplotlib.pyplot as plt
sst_file = "Harvey_SST.txt"
# Read two-column text file: DOY and SST
data = np.loadtxt(sst_file)
DOY = data[:, 0]
SST = data[:, 1]

# Full SST plot
plt.figure()
plt.plot(DOY, SST, color="blue", linewidth=2)
plt.xlabel("Day of Year")
plt.ylabel("SST (°C)")
plt.title("Sea Surface Temperature: Full Record")
plt.grid(True)
plt.savefig("OCNG470_Lab01_SST_Full.pdf")
plt.close()

# Zoomed SST plot
plt.figure()
plt.plot(DOY, SST, color="blue", linewidth=2)
plt.xlabel("Day of Year")
plt.ylabel("SST (°C)")
plt.title("SST: Zoomed, Day 190 to Day 280")
plt.axis([190, 280, 22, 32])
plt.grid(True)
plt.savefig("OCNG470_Lab01_SST_Zoom.pdf")
plt.close()
```
